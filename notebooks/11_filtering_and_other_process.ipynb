```python
import pandas as pd
```

```python
data_path = "../data/bills.csv"
data = pd.read_csv(data_path)
```

```python
def factura_priority_filter(
    df: pd.DataFrame,
    factura_col: str = "FACTURA",
    type_col: str = "EXTRACTION",
):
    priority = {
        "parser": 0,
        "IA+parser": 1,
        "IA": 2,
    }

    is_dup = df.duplicated(subset=factura_col, keep=False)

    prio = df[type_col].map(priority).fillna(99)

    best_prio = prio.groupby(df[factura_col]).transform("min")

    mask = is_dup & (prio == best_prio)

    return mask
```

```python
df_filtered = data[factura_priority_filter(data)]
```

```python
df_filtered.to_excel("../data/bills_filtered.xlsx", index=False)
```
